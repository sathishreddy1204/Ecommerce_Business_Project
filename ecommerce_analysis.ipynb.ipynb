```python
#Importing Libraries

import pandas as pd
import numpy as np
```

```python
#Reading the CSV files

df=pd.read_csv("Ecommerce Project.csv")
```

```python
df
```

```python
#Adding Extra Columns

df['Cost_price'] = round(df['Sales'] * 0.7, 2)
df['Profit'] = round(df['Sales'] - df['Cost_price'], 2)
df['Discount'] = (df['Sales']*0.10).astype(int)
df['Shipping_cost'] = (df['Sales'] * 0.08).astype(int)
np.random.seed(42)
df['Delivery_days'] = np.random.randint(1,11,size=len(df))
df['Customer_rating'] = 3+(np.random.randint(1,11,size=len(df)))*0.2
df['Payment_method'] = np.random.choice(['UPI', 'Credit Card', 'Debit Card', 'COD', 'Net Banking'],size=len(df))
df['Order_status'] = np.random.choice(['Complete','Pending','Cancelled'],size=len(df))
df
```

Data Visulazation using EDA

```python
import matplotlib.pyplot as plt
import seaborn as sns

#Plots using Matplotlib

#Line Chart

Sub_Category = df.groupby('Sub-Category')['Sales'].sum()

plt.figure(figsize=(18,6))
plt.plot(Sub_Category,marker='o')
plt.title("Sub-Category by Sales")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")
plt.show()
```

```python
#BarPlot

State  = df.groupby('State')['Profit'].sum()
short_names = [state[:3] for state in State.index]
plt.figure(figsize=(20,12))
plt.bar(short_names,State.values)
plt.title("State by Profit")
plt.xlabel("States")
plt.ylabel("Profit")

plt.show()
```

```python
#Histogram

plt.figure(figsize=(10,5))
plt.hist(df['Delivery_days'],bins=20)
plt.title("Delivery_day Distribution")
plt.xlabel("Delivery_days")
plt.ylabel("Frequency")
```

```python
#seaborn

#CountPlot

plt.figure(figsize=(6,4))
sns.countplot(x='Region',data=df)
plt.title("Orders by Region")
plt.show()
```

```python
#BoxPlot

plt.figure(figsize=(8,4))
sns.boxplot(x=df['Customer_rating'])
plt.title("Customer_rating Outliers")
plt.show()
```

```python
#Heatmap

plt.figure(figsize=(8,4))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
